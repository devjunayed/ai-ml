# Filtering Based on string

```python
import pandas as pd

data = {
    "Name" : ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah", "Jack David"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "Houston", "Yo Los Ying"],
    "Department" : ["HR", "IT", "Finance", "IT", "HR", "Marketing", "finance", "HR", "Jobless"],
    "Salary" : [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000, 0]
}

df = pd.DataFrame(data)
df
```

## Findings row which has "New" in City (Case Sensitive)

```python

df.loc[df['City'].str.contains("New")]
```

## Findings row which has "New" in City (Non Case Sensitive)

```python
df.loc[df['City'].str.contains("new", case=False)]
```

## Findings row which start with "Los"

```python
df.loc[df['City'].str.contains(r"^Los")]
```

## End with "rk"

```python
df.loc[df['City'].str.contains(r"rk$")]
```

## name starts with a vowel

```python
df.loc[df['Name'].str.contains(r"^[AEIOU]")]
```

## string contains new or los

```python
df.loc[df['City'].str.contains(r"New|Los")]
```

# Handling Duplicate values

```python
import pandas as pd
data = {
    "Name" : ["Alice", "Bob", "Charlie", "Alice", "David", "Bob"],
    "City": ["New York", "London", "Paris", "New York", "Tokyo", "London"],
    "Score" : [85, 90, 78, 80, 95, 90]
}

df = pd.DataFrame(data)
df
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

## Dropping Duplicates 

```python
# Dropping duplicates doesn't perform replace use inplace="True"
df.drop_duplicates()
```

```python
# deleting permanantly
df.drop_duplicates(inplace=True)
df
```

## Deleting duplicates based on column (name)

```python
# Temporarily delete using subset
df.drop_duplicates(subset=['Name', 'City'])
```

```python
df
```

## Selecting which duplicates values to keep

```python
df.drop_duplicates(subset=['Name'], keep="last")
```
